```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('/content/Advertising.csv')
data.head()
```

```python
data = data.iloc[:,1:]
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data.columns
```

```python
cols =['TV', 'Radio', 'Newspaper']
for col in cols:
  plt.plot(data[col],data['Sales'],'ro')
  plt.title('Sales vs {}'.format(col))
  plt.xlabel(col)
  plt.ylabel('Sales')
  plt.show()
```
